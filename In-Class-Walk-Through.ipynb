```python
def summarize_grid(m, verbose=False):
    """Calculates summary statistics for a two-dimensional list of numbers.

    Analyzes a list of sublists containing numbers and calculates overall cell 
    count, total sum, average, and row/column coordinates for the maximum 
    and minimum values.

    Args:
        m (list): A list of lists containing numerical values (ints or floats).

    Returns:
        dict: A dictionary containing summary statistics with the following keys:
            - 'total_cells' (int): Total count of elements across all sublists.
            - 'sum_of_cells' (int or float): Total sum of all numerical values.
            - 'average_of_cells' (float): Arithmetic average of all values.
            - 'max_index' (tuple): Coordinates (row, column) of the largest value.
            - 'min_index' (tuple): Coordinates (row, column) of the smallest value.
    """

    # Create the empty dictionary to populate and send back

    ret_dict = {
        'total_cells': 0,
        'sum_of_cells': 0,
        'max_value': m[0][0],
        'min_value': m[0][0],
        'max_index': (0, 0),
        'min_index': (0, 0),
        'avgerage_of_cells':0}

    for row in range(len(m)):
        for col in range(len(m[row])):
            # print(f"{m[row][col]} - ")
            ret_dict['total_cells'] += 1
            ret_dict['sum_of_cells'] += m[row][col]
            if m[row][col] > ret_dict['max_value']:
                ret_dict['max_value'] = m[row][col]
                ret_dict['max_index'] = (row, col)
            if m[row][col] < ret_dict['min_value']:
                ret_dict['min_value'] = m[row][col]
                ret_dict['min_index'] = (row, col)

    ret_dict['avgerage_of_cells'] = \
        ret_dict['sum_of_cells'] / ret_dict['total_cells']

    if verbose:
        print(ret_dict)
    
    return ret_dict
            

my_summary = summarize_grid(data, verbose=True)
my_summary
```
